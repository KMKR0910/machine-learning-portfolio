Conditional Statements

```python
age=int(input('Enter your age:'))


if age>=25 and age<=70:
   print('Age is valid')
elif age>=18:
   print('Age is not valid')
else:
   print('Error')
```

Loops

1. For Loops

```python
model_prediction = ['Car','Van','Bus','Truck']

for vehicle in model_prediction:
   print (vehicle)

```

2. For Loop with range()

```python
number=10

for i in range (1,number+5,3):
    print(i)
```

For Loops with enumerate()

```python
rain_predictions=['Heavy','Mid','Little']

for index, lable in enumerate(rain_predictions):
   print(f'{index}={lable}')

```

While Loops

```python
number=input('Enter number')

while number<=25:
   print(f'Number is valid {number}')
```

Break & Continue

```python
values=[45,23,56,'Nan',89]

for value in values:
   if value is 'Nan':
      continue

   if value<=50:
      print('High Value')
      break

   print(f'Valid Number:{value}')

```

Exception handling

```python
value='67'

try:
   formatted_value= float(value)
   print(f'Floated Value:{formatted_value:.1%}')

except ValueError:
   print('Error')
finally:
   print('Finished')
```
