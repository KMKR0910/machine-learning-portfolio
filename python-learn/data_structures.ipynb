Lists

Used for ordered collection of data and it is mutable

```python
data= [45,89,'pqr',67,56,25,67,90]

data.append(78)
print(data[2])
print(len(data))

data[2]=89
print(data)
```

List slicing

```python
print(data[:3])
print(data[3:])
print(data[::2])
```

Lists extend, copy, sort, clear ,delete

```python
list_1=[2,4,5]
list_2=[6,8,9]

list_1.extend(list_2)

list_1.pop(1)
list_1.sort()
list_1.sort(reverse=True)

copied_list_1= list_1.copy()
copied_list_2= list(list_1)

print(list_1)
print(list_2)
print(copied_list_1)
print(copied_list_2)

del copied_list_2[3]
copied_list_1.clear()

print(copied_list_1)
print(copied_list_2)
```

3.Sets

Used for unordered collection of data and it is mutable . specaiily used for store unique values

```python
data_set={'car','van','bike','car','bike'}

print(set(data_set))


```

```python
data_set.add('bicycle')
data_set.update('bike','bikes')
print(data_set)
data_set.remove('van')
print(data_set)
```

4.Dictionaries

Used for insertion data and it is mutable. Specaily in AI/Ml used for key value mapping

```python
data_dict={
   'id':'001',
   'name':'kasun',
   'age':'24'

}

print(data_dict)
print(data_dict['name'])
print(data_dict.keys())
print(data_dict.values())
print(data_dict.items())

```
